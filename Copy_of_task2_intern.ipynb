```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style='whitegrid')

df = pd.read_csv('/content/cafe_sales_cleaned.csv', parse_dates=['Transaction Date'])
df['M'] = df['Transaction Date'].dt.month
print(df.shape)
df.head()
```

```python
print(df['Total Spent'].describe())
print('Total revenue:', df['Total Spent'].sum())
print('Skew:', df['Total Spent'].skew())
df[['Quantity','Price Per Unit','Total Spent']].corr()
```

```python
it = df.groupby('Item').agg(rev=('Total Spent','sum'), n=('Item','size')).sort_values('rev', ascending=False)
it['share_%'] = it.rev / it.rev.sum() * 100
print(it)

fig, ax = plt.subplots(1, 2, figsize=(11,4))
ax[0].barh(it.index[::-1], it.rev[::-1], color='#2f5d50'); ax[0].set_title('Revenue by item ($)')
it2 = it.sort_values('n', ascending=False)
ax[1].barh(it2.index[::-1], it2.n[::-1], color='#d98c3f'); ax[1].set_title('Transactions by item')
plt.tight_layout(); plt.show()
```

```python
print(df.groupby('Item')['Quantity'].mean())
sns.heatmap(df[['Quantity','Price Per Unit','Total Spent']].corr(), annot=True, fmt='.2f', cmap='YlGn')
plt.title('Correlation matrix'); plt.show()
```

```python
mo = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
days = pd.Series([31,28,31,30,31,30,31,31,30,31,30,31], index=range(1,13))
m = df.groupby('M')['Total Spent'].sum()

fig, ax = plt.subplots(1, 2, figsize=(11,4))
ax[0].plot(mo, m.values, marker='o', color='#2f5d50'); ax[0].axhline(m.mean(), ls='--', color='gray')
ax[0].set_title('Monthly revenue ($)')
ax[1].bar(mo, (m/days).values, color='#d98c3f'); ax[1].set_ylim(190,240)
ax[1].set_title('Average revenue per day ($)')
plt.tight_layout(); plt.show()
```

```python
wd = ['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
w = df.groupby('Weekday')['Total Spent'].sum().reindex(wd)
nd = df.groupby('Weekday')['Transaction Date'].nunique().reindex(wd)
(w/nd).plot(kind='bar', color='#2f5d50', figsize=(7,4), ylim=(180,250))
plt.title('Average daily revenue by weekday ($)'); plt.show()
```

```python
d = df.groupby('Transaction Date')['Total Spent'].sum()
z = (d - d.mean()) / d.std()

plt.figure(figsize=(11,4))
plt.plot(d.index, d.values, color='#bbb', lw=.8, label='Daily')
plt.plot(d.index, d.rolling(7).mean(), color='#2f5d50', lw=2, label='7-day avg')
plt.scatter(d.index[z.abs()>2.5], d[z.abs()>2.5], color='crimson', zorder=5, label='Outliers (|z|>2.5)')
plt.title('Daily revenue ($)'); plt.legend(); plt.show()

print(z[z.abs()>2.5])
print(d.nlargest(3)); print(d.nsmallest(3))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11,4))
sns.histplot(df['Total Spent'], bins=25, color='#2f5d50', ax=ax[0]); ax[0].set_title('Transaction value')
sns.boxplot(data=df, x='Item', y='Total Spent', order=it.index, color='#9fc2b6', ax=ax[1])
ax[1].tick_params(axis='x', rotation=45); ax[1].set_title('Value by item')
plt.tight_layout(); plt.show()

print((df['Total Spent'] >= 20).sum(), 'orders >= $20')
print(df[df['Total Spent'] >= 20]['Item'].value_counts())
```

## Key Insights

**1. Three items generate 55% of revenue.**
Salad ($18,220, 22.8%), Sandwich and Smoothie together earn 55% of total revenue. Coffee is the most ordered item (1,241 transactions) but contributes only 9.4%. Action: protect stock and promotion of the top earners and use cheap items (coffee, cookies) as add-ons.

**2. Price drives revenue, not quantity.**
Average quantity is about 3 units for every item (2.98 to 3.05), and quantity is uncorrelated with price (r = 0.006). Total Spent correlates with quantity (0.70) and price (0.65). Action: test bundles and upsell toward higher-priced items.

**3. Sales are stable all year with no strong seasonality.**
Monthly revenue ranges from $6,178 (Feb) to $7,040 (Oct), and February is low only because it has 28 days. Per day, it earns $220.6 against an average of $219.3. Action: plan staffing and stock on a flat baseline; growth must come from pricing and product mix.

**4. Wednesday is the weakest day, and weekends are not busier.**
Wednesday averages $208.9 per day versus $225.5 on Thursday, a gap of about 8%. Saturday and Sunday are not higher than weekdays. Action: run a mid-week promotion and review weekend hours and offers.

**5. Daily revenue is volatile with a few anomalies.**
Daily revenue ranges from $101.5 to $371.5 (std $53.7). Three days are above 2.5 standard deviations: 24 Jul ($371.5), 5 Jan ($360.5) and 6 Jun ($355.5). The 7-day average stays flat, so these are one-day spikes, not trends. Action: check whether these days had events or promotions and repeat what worked.

**6. High-value orders are rare and come from three items.**
Transaction value is right-skewed (median $7.50, mean $8.83). Only 963 orders (10.6%) are $20 or more, and all are Salad (498), Sandwich (240) or Smoothie (225). Action: promote these items in combos to raise the average bill.

**7. Data quality: many Unknown values.**
31.7% of payment methods (2,870) and 39.6% of locations (3,592) are Unknown, and 1,169 rows have both missing. Among known values, the split is even. Action: make these fields mandatory in the billing system so channel analysis is reliable.
